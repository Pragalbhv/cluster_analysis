```python
def shell_cutoffs(rdfs, densities, pair="Cl-Pu", r_max=10.0, min_prominence=0.):
    """
    Calculate shell cutoffs based on coordination number and radial distance.
    
    This function integrates RDF to get coordination number vs r, then uses
    derivatives (similar to ovito_bond_shell_analysis.ipynb) to detect shell boundaries.
    
    Args:
        rdfs: Dictionary containing RDF data with keys like 'r' and 'Cl-Pu'
        densities: Dictionary of species densities {species: density}
        pair: RDF pair key (default: "Cl-Pu" for Pu-Cl coordination)
        r_max: Maximum integration distance (default: 10.0 Å)
        min_prominence: Minimum prominence for peak detection in derivatives (default: 0.1)
    
    Returns:
        numpy.ndarray: Array of shell cutoff radii (in Å) where shell boundaries are detected
    """
    # Extract radial distance and RDF values for the specified pair
    r = rdfs["r"]
    g_r = rdfs[pair]
    
    # Determine which species is the neighbor (for density lookup)
    # For "Cl-Pu", we want Cl density (neighbor species)
    neighbor_species = pair.split("-")[0]  # "Cl" for "Cl-Pu"
    rho_j = densities[neighbor_species]
    
    # Calculate coordination number vs r by integrating RDF
    r_cn, cn_values = calculate_coordination_number_vs_r(
        r, g_r, rho_j, r_max=r_max
    )
    
    # Calculate first derivative of coordination number
    first_deriv = np.gradient(cn_values, r_cn)
    
    # Detect shell boundaries using proper pattern (derivative peaks then goes to zero)
    shell_indices = detect_proper_shells(first_deriv, min_prominence=min_prominence)
    
    # Convert indices to actual r values (shell cutoffs)
    if len(shell_indices) > 0:
        shell_cutoffs = r_cn[shell_indices]
    else:
        shell_cutoffs = np.array([])
    
    return shell_cutoffs
```

```python
# Test the shell_cutoffs function with current data
print("=" * 60)
print("SHELL CUTOFFS CALCULATION")
print("=" * 60)

# Calculate shell cutoffs for Pu-Cl coordination
shell_cutoffs_pucl = shell_cutoffs(
    rdfs=current_rdf,
    densities=current_densities,
    pair="Cl-Pu",
    r_max=rmax,
    min_prominence=0.1
)

print(f"\nDetected {len(shell_cutoffs_pucl)} shell boundary(ies) for Pu-Cl coordination:")
if len(shell_cutoffs_pucl) > 0:
    for i, cutoff in enumerate(shell_cutoffs_pucl, 1):
        print(f"  Shell {i} cutoff: {cutoff:.3f} Å")
else:
    print("  No shell boundaries detected with current parameters.")
    print("  Try adjusting min_prominence or r_max if needed.")

# Compare with the current cutoff
print(f"\nCurrent Pu-Cl cutoff from RDF minimum: {current_cutoffs[('Pu', 'Cl')]:.3f} Å")
```

```python
# Calculate Pu-Cl cutoffs for all compositions
print("=" * 80)
print("PU-CL SHELL CUTOFFS ACROSS COMPOSITIONS")
print("=" * 80)

# Base directory and temperature (using same temperature as current example)
base_dir = "/pscratch/sd/p/pvashi/irp/irp_mace_l_2/irp/density/NaCl-PuCl3"
temperature = "T1100K"  # Using same temperature as current example

# Composition values (x values)
compositions = ["0.00", "0.07", "0.17", "0.27", "0.40", "0.50", "0.60", "0.67", "0.75", "0.88", "1.00"]

# Storage for results
results = []

# Parameters for shell cutoff calculation
r_max = 10.0
min_prominence = 0.1
rdf_samples = 100

print(f"\nProcessing {len(compositions)} compositions at {temperature}...")
print(f"Parameters: r_max={r_max} Å, min_prominence={min_prominence}, rdf_samples={rdf_samples}\n")

for x_val in compositions:
    # Construct file path
    infile_path = f"{base_dir}/x{x_val}/{temperature}/dump.lammpstrj"
    
    print(f"Processing x={x_val}...", end=" ")
    
    try:
        # Check if file exists
        if not os.path.exists(infile_path):
            print(f"SKIPPED (file not found)")
            results.append({
                'x': float(x_val),
                'cutoff_shell': None,
                'cutoff_rdf_min': None,
                'status': 'file_not_found'
            })
            continue
        
        # Load pipeline
        pipe = import_file(infile_path, multiple_frames=True)
        
        # Compute RDFs
        result = canonical_cluster_workflow_rdfs_only(
            pipeline=pipe,
            rdf_samples=rdf_samples
        )
        
        # Get RDF data and densities
        rdf_data = result["rdf"]
        rdf_cutoffs = result["pair_cutoffs"]
        densities = get_species_densities_from_data(pipe.compute(pipe.source.num_frames-1))
        
        # Calculate shell cutoffs using derivative method
        shell_cutoffs_pucl = shell_cutoffs(
            rdfs=rdf_data,
            densities=densities,
            pair="Cl-Pu",
            r_max=r_max,
            min_prominence=min_prominence
        )
        
        # Get RDF minimum cutoff for comparison
        rdf_min_cutoff = rdf_cutoffs.get(('Pu', 'Cl'), None)
        
        # Store first shell cutoff (if detected)
        first_shell_cutoff = shell_cutoffs_pucl[0] if len(shell_cutoffs_pucl) > 0 else None
        
        results.append({
            'x': float(x_val),
            'cutoff_shell': first_shell_cutoff,
            'cutoff_rdf_min': rdf_min_cutoff,
            'num_shells': len(shell_cutoffs_pucl),
            'all_shells': shell_cutoffs_pucl.tolist() if len(shell_cutoffs_pucl) > 0 else [],
            'status': 'success'
        })
        
        if first_shell_cutoff is not None:
            print(f"✓ Shell cutoff: {first_shell_cutoff:.3f} Å (RDF min: {rdf_min_cutoff:.3f} Å)")
        else:
            print(f"⚠ No shell detected (RDF min: {rdf_min_cutoff:.3f} Å)")
            
    except Exception as e:
        print(f"ERROR: {str(e)}")
        results.append({
            'x': float(x_val),
            'cutoff_shell': None,
            'cutoff_rdf_min': None,
            'status': f'error: {str(e)}'
        })

# Summary table
print("\n" + "=" * 80)
print("SUMMARY: PU-CL CUTOFFS BY COMPOSITION")
print("=" * 80)
print(f"{'Composition (x)':<15} {'Shell Cutoff (Å)':<20} {'RDF Min (Å)':<15} {'Difference (Å)':<15} {'Status':<20}")
print("-" * 80)

for res in results:
    x = res['x']
    shell_cut = res['cutoff_shell']
    rdf_cut = res['cutoff_rdf_min']
    
    if shell_cut is not None and rdf_cut is not None:
        diff = shell_cut - rdf_cut
        shell_str = f"{shell_cut:.3f}"
        rdf_str = f"{rdf_cut:.3f}"
        diff_str = f"{diff:+.3f}"
    else:
        shell_str = "N/A" if shell_cut is None else f"{shell_cut:.3f}"
        rdf_str = "N/A" if rdf_cut is None else f"{rdf_cut:.3f}"
        diff_str = "N/A"
    
    status = res['status']
    if status == 'success' and shell_cut is None:
        status = 'no_shell_detected'
    
    print(f"{x:<15.2f} {shell_str:<20} {rdf_str:<15} {diff_str:<15} {status:<20}")

# Create visualization
print("\n" + "=" * 80)
print("VISUALIZATION: PU-CL CUTOFFS VS COMPOSITION")
print("=" * 80)

# Extract data for plotting
x_values = [r['x'] for r in results]
shell_cutoffs_plot = [r['cutoff_shell'] if r['cutoff_shell'] is not None else np.nan for r in results]
rdf_cutoffs_plot = [r['cutoff_rdf_min'] if r['cutoff_rdf_min'] is not None else np.nan for r in results]

# Create plot
fig, ax = plt.subplots(figsize=(12, 7))

# Plot shell cutoffs
valid_shell = ~np.isnan(shell_cutoffs_plot)
if np.any(valid_shell):
    ax.scatter([x_values[i] for i in range(len(x_values)) if valid_shell[i]], 
               [shell_cutoffs_plot[i] for i in range(len(shell_cutoffs_plot)) if valid_shell[i]],
               color='green', s=100, marker='o', label='Shell Cutoff (Derivative Method)', zorder=5)
    ax.plot([x_values[i] for i in range(len(x_values)) if valid_shell[i]], 
            [shell_cutoffs_plot[i] for i in range(len(shell_cutoffs_plot)) if valid_shell[i]],
            color='green', linestyle='--', alpha=0.5, linewidth=2)

# Plot RDF minimum cutoffs
valid_rdf = ~np.isnan(rdf_cutoffs_plot)
if np.any(valid_rdf):
    ax.scatter([x_values[i] for i in range(len(x_values)) if valid_rdf[i]], 
               [rdf_cutoffs_plot[i] for i in range(len(rdf_cutoffs_plot)) if valid_rdf[i]],
               color='red', s=100, marker='s', label='RDF Minimum Cutoff', zorder=5)
    ax.plot([x_values[i] for i in range(len(x_values)) if valid_rdf[i]], 
            [rdf_cutoffs_plot[i] for i in range(len(rdf_cutoffs_plot)) if valid_rdf[i]],
            color='red', linestyle='--', alpha=0.5, linewidth=2)

ax.set_xlabel('Composition (x)', fontsize=12, fontweight='bold')
ax.set_ylabel('Pu-Cl Cutoff (Å)', fontsize=12, fontweight='bold')
ax.set_title(f'Pu-Cl Cutoffs vs Composition at {temperature}', fontsize=14, fontweight='bold')
ax.legend(fontsize=11, framealpha=0.9, loc='best')
ax.grid(True, alpha=0.3)
ax.set_facecolor('white')

plt.tight_layout()
plt.show()

print(f"\nProcessed {len([r for r in results if r['status'] == 'success'])}/{len(results)} compositions successfully.")
```
